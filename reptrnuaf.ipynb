```python
from pathlib import Path
import pandas as pd
import glob

# ==================================================
# RUTAS
# ==================================================

RUTA_BASE = Path(r"C:\Users\NI38504\Documents\Home\REMESAS\INSUMOS\Julio")
RUTA_BASE = Path(r"\\172.25.16.15\Reportes Cobis_BI\Remesas\2026\Julio")

RUTA_SALIDA = Path(
    r"C:\Users\NI38504\Documents\Home\REMESAS\INSUMOS\Julio\CONSOLIDADO"
)

RUTA_SALIDA.mkdir(exist_ok=True)

ARCHIVO_SALIDA = (
    RUTA_SALIDA / "Consolidado_Reptrnuaf.xlsx"
)

# ==================================================
# ACUMULADOR
# ==================================================

dfs = []

# ==================================================
# RECORRER CARPETAS
# ==================================================

print(RUTA_BASE)

for carpeta in RUTA_BASE.iterdir():

    if not carpeta.is_dir():
        continue

    archivos = glob.glob(
        str(carpeta / "reptrnuaf*.xls")
    )

    if not archivos:
        continue

    for archivo in archivos:

        try:

            print(f"Procesando: {archivo}")

            df = pd.read_csv(
                archivo,
                sep="\t",
                dtype=str,
                encoding="latin-1",
                engine="python"
            )

            df.columns = [
                str(col).strip()
                for col in df.columns
            ]

            df["CARPETA_ORIGEN"] = carpeta.name

            dfs.append(df)

        except Exception as e:

            print(
                f"Error procesando {archivo}"
            )
            print(str(e))

# ==================================================
# CONSOLIDAR
# ==================================================

if dfs:

    df_final = pd.concat(
        dfs,
        ignore_index=True
    )

    with pd.ExcelWriter(
        ARCHIVO_SALIDA,
        engine="openpyxl"
    ) as writer:

        df_final.to_excel(
            writer,
            sheet_name="DEPOSITOS A CUENTA",
            index=False
        )

    print("\nProceso finalizado")
    print(f"Archivo generado: {ARCHIVO_SALIDA}")
    print(f"Registros: {len(df_final):,}")

else:

    print(
        "No se encontraron archivos reptrnuaf para procesar."
    )
```
